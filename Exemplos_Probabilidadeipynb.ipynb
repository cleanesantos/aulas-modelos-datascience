# Aplicação computacional: distribuições discretas

Situação-problema
Uma equipe de software está analisando o comportamento de um sistema durante testes. Considere que:

```python
from scipy.stats import randint, bernoulli, binom, poisson
```


## 1. Um servidor é escolhido aleatoriamente entre 6 servidores disponíveis. Qual a probabilidade de selecionar o servidor 4?


```python
# 1. Uniforme discreta
p_servidor_4 = randint.pmf(4, low=1, high=7)
print("Servidor 4:", p_servidor_4)
```

##2. Uma requisição tem 30% de chance de apresentar sucesso em uma determinada condição. Qual a probabilidade de sucesso?


```python
# 2. Bernoulli
p_sucesso = bernoulli.pmf(1, p=0.30)
print("Sucesso:", p_sucesso)
```

##3. Em 12 requisições independentes, cada uma tem probabilidade de 6,5% de apresentar erro. Qual a probabilidade de ocorrerem exatamente 2 erros?

```python
# 3. Binomial
p_2_erros = binom.pmf(2, n=12, p=0.065)
print("Exatamente 2 erros:", p_2_erros)
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)

# 1. Binomial
# Número de erros em 20 requisições
erros = np.random.binomial(n=20, p=0.10, size=200)

plt.hist(erros, bins=range(erros.min(), erros.max()+2), edgecolor="black")
plt.xlabel("Número de erros")
plt.ylabel("Frequência")
plt.title("Erros em 20 requisições")
plt.show()
```

##4. Em média, ocorrem 2 falhas por hora. Qual a probabilidade de ocorrer exatamente 1 falha em uma hora?

```python
# 4. Poisson
p_1_falha = poisson.pmf(1, mu=2)
print("Exatamente 1 falha:", p_1_falha)
```

# Aplicação computacional: distribuições Contínuas

Situação-problema
Uma equipe de software está analisando tempos e medidas de desempenho de um sistema:

##Uniforme contínua: o tempo de uma determinada operação pode assumir qualquer valor entre 0 e 20 segundos, com igual densidade de probabilidade. Qual a probabilidade de o tempo ficar entre 5 e 12 segundos?

```python
from scipy.stats import uniform, expon, norm

# 1. Uniforme contínua
p_uniforme = (
    uniform.cdf(12, loc=0, scale=20)
    - uniform.cdf(5, loc=0, scale=20)
)
print("Entre 5 e 12 segundos:", p_uniforme)
```

##Exponencial: ocorrem, em média, 4 eventos por hora. Qual a probabilidade de o tempo até o próximo evento ser superior a 30 minutos?

```python
# 2. Exponencial
p_mais_30min = expon.sf(0.5, scale=1/4)

print("Mais de 30 minutos:", p_mais_30min)
```

```python
# 3. Exponencial
# Tempo até a próxima falha, em horas
tempos_falha = np.random.exponential(scale=2, size=300)

plt.hist(tempos_falha, bins=15, density=True, edgecolor="black")
plt.xlabel("Tempo até a próxima falha (horas)")
plt.ylabel("Densidade")
plt.title("Tempo até a próxima falha")
plt.show()
```

##Normal: o tempo de execução de uma aplicação segue aproximadamente uma distribuição Normal com média de 2,8 segundos e desvio padrão de 1,2 segundo. Qual a probabilidade de o tempo ultrapassar 4 segundos?

```python
# 3. Normal
p_maior_4 = norm.sf(4, loc=2.8, scale=1.2)
print("Acima de 4 segundos:", p_maior_4)
```

```python
# 2. Normal
# Tempos de execução em milissegundos
tempos = np.random.normal(loc=100, scale=15, size=300)

plt.hist(tempos, bins=15, density=True, edgecolor="black")
plt.xlabel("Tempo de execução (ms)")
plt.ylabel("Densidade")
plt.title("Distribuição dos tempos de execução")
plt.show()
```
