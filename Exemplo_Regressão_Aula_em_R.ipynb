```r
# =========================================================
# EXEMPLO 1
# Quilometragem do carro x preço de venda
# =========================================================

# Dados
x <- c(40, 30, 30, 25, 50, 60, 65, 10, 15, 20, 55, 40, 35, 30)

y <- c(1000, 1500, 1200, 1800, 800, 1000, 500,
       3000, 2500, 2000, 800, 1500, 2000, 2000)

# Criando o modelo de regressão linear
modelo1 <- lm(y ~ x)

# Visualizar resultado completo
summary(modelo1)

# Coeficientes da reta
b0 <- coef(modelo1)[1]
b1 <- coef(modelo1)[2]

cat("b0 =", b0, "\n")
cat("b1 =", b1, "\n")

# R²
r2 <- summary(modelo1)$r.squared

cat("R² =", r2, "\n")

# Correlação
r <- cor(x, y)

cat("Correlação =", r, "\n")
```

```r
# ============================================================
# EXEMPLO 2
# INVESTIMENTO EM PUBLICIDADE x FATURAMENTO
# ============================================================

# ------------------------------------------------------------
# 1. DADOS
# Valores em milhares de reais
# ------------------------------------------------------------

x <- c(1, 2, 3, 4, 5)

y <- c(10, 13, 15, 14, 18)


# ------------------------------------------------------------
# 2. TABELA AUXILIAR
# ------------------------------------------------------------

tabela2 <- data.frame(
  Campanha = 1:length(x),
  X = x,
  Y = y,
  XY = x * y,
  X2 = x^2,
  Y2 = y^2
)

tabela2


# ------------------------------------------------------------
# 3. SOMATÓRIOS
# ------------------------------------------------------------

n <- length(x)

soma_x  <- sum(x)
soma_y  <- sum(y)
soma_xy <- sum(x * y)
soma_x2 <- sum(x^2)
soma_y2 <- sum(y^2)

cat("n =", n, "\n")
cat("Soma X =", soma_x, "\n")
cat("Soma Y =", soma_y, "\n")
cat("Soma XY =", soma_xy, "\n")
cat("Soma X² =", soma_x2, "\n")
cat("Soma Y² =", soma_y2, "\n")


# ------------------------------------------------------------
# 4. MÉDIAS
# ------------------------------------------------------------

media_x <- mean(x)
media_y <- mean(y)

cat("\nMédia de X =", media_x, "\n")
cat("Média de Y =", media_y, "\n")


# ------------------------------------------------------------
# 5. CÁLCULO MANUAL DE b1
# ------------------------------------------------------------

b1_manual <- (
  soma_xy - n * media_x * media_y
) / (
  soma_x2 - n * media_x^2
)

cat("\nb1 calculado manualmente =", b1_manual, "\n")


# ------------------------------------------------------------
# 6. CÁLCULO MANUAL DE b0
# ------------------------------------------------------------

b0_manual <- media_y - b1_manual * media_x

cat("b0 calculado manualmente =", b0_manual, "\n")


# ------------------------------------------------------------
# 7. EQUAÇÃO DA RETA
# ------------------------------------------------------------

cat(
  "\nEquação estimada:\n",
  "Y estimado =",
  round(b0_manual, 2),
  "+",
  round(b1_manual, 2),
  "* X\n"
)


# ------------------------------------------------------------
# 8. REGRESSÃO LINEAR COM lm()
# ------------------------------------------------------------

modelo2 <- lm(y ~ x)

summary(modelo2)


# ------------------------------------------------------------
# 9. COEFICIENTES CALCULADOS PELO R
# ------------------------------------------------------------

b0_r <- coef(modelo2)[1]
b1_r <- coef(modelo2)[2]

cat("\nb0 pelo lm() =", b0_r, "\n")
cat("b1 pelo lm() =", b1_r, "\n")


# ------------------------------------------------------------
# 10. COEFICIENTE DE DETERMINAÇÃO R²
# ------------------------------------------------------------

r2 <- summary(modelo2)$r.squared

cat("\nR² =", r2, "\n")


# ------------------------------------------------------------
# 11. COEFICIENTE DE CORRELAÇÃO
# ------------------------------------------------------------

r <- cor(x, y)

cat("Correlação r =", r, "\n")


# ------------------------------------------------------------
# 12. CONFERÊNCIA
# Em regressão linear simples: R² = r²
# ------------------------------------------------------------

cat("\nr² =", r^2, "\n")
cat("R² =", r2, "\n")


# ------------------------------------------------------------
# 13. GRÁFICO DE DISPERSÃO + RETA DE REGRESSÃO
# ------------------------------------------------------------

plot(
  x,
  y,
  main = "Investimento em publicidade x Faturamento",
  xlab = "Investimento em publicidade (R$ mil)",
  ylab = "Faturamento (R$ mil)",
  pch = 19,
  xlim = c(1, 6),
  ylim = c(8, 20)
)

abline(
  modelo2,
  lwd = 2
)

legend(
  "topleft",
  legend = c(
    paste("r =", round(r, 3)),
    paste("R² =", round(r2, 3))
  ),
  bty = "n"
)


# ------------------------------------------------------------
# 14. NOVO CASO
# Nova campanha com investimento de R$ 6 mil
# ------------------------------------------------------------

nova_campanha <- data.frame(
  x = 6
)

previsao <- predict(
  modelo2,
  newdata = nova_campanha
)

cat(
  "\nInvestimento da nova campanha: R$ 6 mil\n"
)

cat(
  "Faturamento estimado:",
  round(previsao, 2),
  "mil reais\n"
)


# ------------------------------------------------------------
# 15. ADICIONANDO A NOVA PREVISÃO AO GRÁFICO
# ------------------------------------------------------------

plot(
  x,
  y,
  main = "Regressão linear e previsão de uma nova campanha",
  xlab = "Investimento em publicidade (R$ mil)",
  ylab = "Faturamento (R$ mil)",
  pch = 19,
  xlim = c(1, 6.5),
  ylim = c(8, 21)
)

abline(
  modelo2,
  lwd = 2
)

# Ponto correspondente à nova previsão
points(
  6,
  previsao,
  pch = 17,
  cex = 1.4
)

# Identificação do ponto previsto
text(
  6,
  previsao,
  labels = paste(
    "Previsto =",
    round(previsao, 1)
  ),
  pos = 2
)

legend(
  "topleft",
  legend = c(
    paste("r =", round(r, 3)),
    paste("R² =", round(r2, 3))
  ),
  bty = "n"
)
```
